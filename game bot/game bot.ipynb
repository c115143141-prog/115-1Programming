```python
import ctypes
import time

# =========================
# Windows API
# =========================
user32 = ctypes.windll.user32
gdi32 = ctypes.windll.gdi32

# =========================
# 偵測設定
# =========================
X_START = 1100
X_END = 1200

Y_TOP = 485
Y_BOTTOM = 523

TARGET_COLOR = (83, 83, 83)

# =========================
# 時間設定
# =========================
JUMP_COOLDOWN = 0.1
DOWN_TIME = 0.5

# =========================
# 取得螢幕像素顏色
# =========================
def get_pixel(x, y):
    hdc = user32.GetDC(0)
    color = gdi32.GetPixel(hdc, x, y)
    user32.ReleaseDC(0, hdc)

    r = color & 0xFF
    g = (color >> 8) & 0xFF
    b = (color >> 16) & 0xFF

    return (r, g, b)

# =========================
# 偵測範圍
# =========================
def detect():
    for x in range(X_START, X_END + 1):
        for y in range(Y_TOP, Y_BOTTOM + 1):

            if get_pixel(x, y) == TARGET_COLOR:
                return True

    return False

# =========================
# 按鍵
# =========================
def press_space():
    user32.keybd_event(0x20, 0, 0, 0)
    user32.keybd_event(0x20, 0, 2, 0)

def down_key():
    user32.keybd_event(0x28, 0, 0, 0)

def up_key():
    user32.keybd_event(0x28, 0, 2, 0)

# =========================
# 開始
# =========================
print("3秒內請切換到 chrome://dino ...")
time.sleep(3)

press_space()

last_jump = 0
down_start = 0
downing = False

while True:

    now = time.time()

    # =========================
    # ↓ 按住 0.5 秒
    # =========================
    if downing:

        if now - down_start >= DOWN_TIME:
            up_key()
            downing = False

    # =========================
    # 偵測目標
    # =========================
    if detect():

        if now - last_jump >= JUMP_COOLDOWN:

            # 跳
            press_space()

            last_jump = now

            # 開始按 ↓
            if not downing:
                down_key()
                down_start = now
                downing = True          
```
